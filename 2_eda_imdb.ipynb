### 6. Demuestra que es mejor tener nombres de columnas sin espacios (notación corchetes y notación punto-variable).

#### ¿por qué es mejor no tener espacios en los nombres de columnas?

*Notación con corchetes → df['Nombre Columna']*

*Notación punto-variable → df.nombre_columna*

##### Lo anterior muestra que:

<p> Con espacios, solo funciona la notación con corchetes </p>
<p> Sin espacios, puedes usar ambas, y el código es más limpio y legible </p>

### En el DataFramre tenemos la columna "Revenue (Millions)" 
### --------------------------------------------------------

```python
# Acceso usando corchetes FUNCIONA

df["Revenue (Millions)"].head() 
```

```python
# Acceso usando punto-variable (NO FUNCIONA)
#Esto da error porque Python interpreta el espacio como separación de palabras.

df.Revenue (Millions).head()
```

##### CONCLUSION
*Tener nombres de columnas sin espacios permite usar tanto la notación con corchetes como la notación punto-variable, haciendo el código más legible, limpio y menos propenso a errores. Por esta razón, es una buena práctica reemplazar espacios por guiones bajos en los nombres de columnas.*

### 21. Obten los valores únicos de la variable Rating y la frecuencia total de cada uno de estos valores. Posteriormente crea un nuevo dataframe con essos valores

```python
# Frecuencia de cada valor en la columna 'Rating'
rating_freq = df['rating'].value_counts().reset_index()
# Renombrando las columnas para mayor claridad
rating_freq.columns = ['Rating', 'Frecuencia']
# Mostrando la frecuencia de cada rating
rating_freq
```

### 22. Ordena el nuevo dataframe por la variable `rating`

```python
# Ordenando el DataFrame por la columna 'Rating'
rating_freq = rating_freq.sort_values(by='Rating').reset_index(drop=True)
# Mostrando el DataFrame ordenado
rating_freq
```

### 23. Crea una gráfica de barras con este nuevo dataframe ordenado

```python
# Graficando la frecuencia de ratings
ax = rating_freq.plot(
    x='Rating',
    y='Frecuencia',
    kind='bar',
    color='orange',
    edgecolor='black',
    figsize=(14,7)
)
# Etiquetas y título
plt.xlabel('Rating')
plt.ylabel('Frecuencia')
plt.title('Frecuencia de Ratings')

# Mostrando los valores encima de las barras
for container in ax.containers:
    ax.bar_label(container)

# Mostrar la gráfica
plt.show()
```

### 24. Crea la matriz de correlación del dataframe de películas

```python
corr_matrix = df.corr(numeric_only=True)
corr_matrix

```

### 25. Grafica la matriz de correlación utilizando `matshow()`

```python
# Visualización de la matriz de correlación
# Configuración del tamaño de la figura
plt.figure(figsize=(10, 8))
# Mapa de calor de la matriz de correlación
plt.matshow(corr_matrix, fignum=1, cmap='coolwarm')
# Barra de color para indicar la escala de correlación
plt.colorbar()
# Etiquetas en los ejes
plt.xticks(
    range(len(corr_matrix.columns)),
    corr_matrix.columns,
    rotation=90
)
plt.yticks(
    range(len(corr_matrix.columns)),
    corr_matrix.columns
)

# Etiquetas dentro de cada celda
for i in range(len(corr_matrix.columns)):
    for j in range(len(corr_matrix.columns)):
        value = corr_matrix.iloc[i, j]
        color = 'white' if abs(value) > 0.5 else 'black'
        plt.text(
            j, i,
            f"{value:.2f}",
            ha='center',
            va='center',
            color=color,
            fontsize=9
        )
# Título de la gráfica
plt.title('Matriz de correlación', pad=20)
# Mostrar la gráfica
plt.show()
```
